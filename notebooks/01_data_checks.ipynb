### 7. Ceded ÷ direct premium swings

Ceded ÷ direct is the share of premium passed to reinsurers. If it jumps between AYs, the net triangle mixes different reinsurance setups. Swing = highest ratio − lowest ratio across AYs. AYs with direct premium of 0 get no ratio.

```python
# PPA (check the column names against your headers)
ppa_cede = ppa_known.groupby(["GRCODE", "AccidentYear"])[["EarnedPremDIR", "EarnedPremCeded"]].first()
ppa_cede["ratio"] = (ppa_cede.EarnedPremCeded / ppa_cede.EarnedPremDIR).where(ppa_cede.EarnedPremDIR != 0)
print("company-AYs with direct = 0:", (ppa_cede.EarnedPremDIR == 0).sum())
ppa_swing = ppa_cede.groupby("GRCODE")["ratio"].agg(lambda s: s.max() - s.min())
print(ppa_swing.describe())
print(ppa_swing.sort_values(ascending=False).head(15))
```

```python
# WC (check the column names against your headers)
wc_cede = wc_known.groupby(["GRCODE", "AccidentYear"])[["EarnedPremDIR", "EarnedPremCeded"]].first()
wc_cede["ratio"] = (wc_cede.EarnedPremCeded / wc_cede.EarnedPremDIR).where(wc_cede.EarnedPremDIR != 0)
print("company-AYs with direct = 0:", (wc_cede.EarnedPremDIR == 0).sum())
wc_swing = wc_cede.groupby("GRCODE")["ratio"].agg(lambda s: s.max() - s.min())
print(wc_swing.describe())
print(wc_swing.sort_values(ascending=False).head(15))
```

## Summary of Data Checks

```python
def probe_summary(known):
    k = known.assign(Reported=known.IncurredLosses - known.BulkLoss)
    ay = k.drop_duplicates(["GRCODE", "AccidentYear"])   # premium: one value per company-AY
    tests = {                                            # (cell flag, grouping key)
        "3. Reported < paid": (k.Reported < k.CumPaidLoss, k.GRCODE),
        "4. Negative bulk":   (k.BulkLoss < 0,             k.GRCODE),
        "5. Premium <= 0":    (ay.EarnedPremNet <= 0,      ay.GRCODE),
    }
    out = {name: {"cells": int(m.sum()), "lines": int(m.groupby(key).any().sum())}
           for name, (m, key) in tests.items()}
    out["6. Zero paid"] = {"cells": np.nan,
        "lines": int(((k.CumPaidLoss != 0).groupby(k.GRCODE).sum() == 0).sum())}
    dir0 = ay.EarnedPremDIR == 0
    out["7. Direct = 0 (no ratio)"] = {"cells": int(dir0.sum()),
        "lines": int(dir0.groupby(ay.GRCODE).all().sum())}
    return pd.DataFrame(out).T.astype("Int64")

def swing_summary(known):
    ay = known.drop_duplicates(["GRCODE", "AccidentYear"])
    ratio = (ay.EarnedPremCeded / ay.EarnedPremDIR).where(ay.EarnedPremDIR != 0)
    g = ratio.groupby(ay.GRCODE)
    swing = (g.max() - g.min()).dropna()
    return swing.quantile([0.5, 0.75, 1.0]).rename(index={0.5: "median", 0.75: "p75", 1.0: "max"})

display(pd.concat({"PPA": probe_summary(ppa_known), "WC": probe_summary(wc_known)}, axis=1))
display(
    pd.DataFrame({"PPA": swing_summary(ppa_known), "WC": swing_summary(wc_known)})
      .style.format(precision=2)
      .set_caption("7. Ceded ÷ direct premium swing (max − min across AYs, per company-line)")
)
```

## Check 6. Zero paid
Question: which company-lines have no payments in any known cell and how can a cleaning rule be validated using this exclusion criteria?

### 6.1 Check which company lines report zero payment activity (PPA and WC)

```python
# PPA
ppa_paidcells = (ppa_known.CumPaidLoss != 0).groupby(ppa_known.GRCODE).sum()
ppa_zeropaid = ppa_paidcells[ppa_paidcells == 0]
print("company-lines:", len(ppa_zeropaid))
print(sorted(ppa_zeropaid.index))
```

```python
# WC
wc_paidcells = (wc_known.CumPaidLoss != 0).groupby(wc_known.GRCODE).sum()
wc_zeropaid = wc_paidcells[wc_paidcells == 0]
print("company-lines:", len(wc_zeropaid))
print(sorted(wc_zeropaid.index))
```

### 6.2 Follow-up: premium. check if companies with no payment activity earn premium.

```python
def zero_paid_premium(known):
    # Probe 6 mask: True = company-line has no nonzero paid cell
    mask = (known.CumPaidLoss != 0).groupby(known.GRCODE).sum() == 0
    flagged = mask[mask].index

    # Premium repeats on every lag row of an AY, so keep one row per company-AY
    ay = (known[known.GRCODE.isin(flagged)]
          .drop_duplicates(["GRCODE", "AccidentYear"]))

    return ay.groupby("GRCODE").EarnedPremNet.agg(
        mean_prem="mean",
        total_prem="sum",
        AYs_with_prem=lambda s: (s > 0).sum(),
    )

for name, known in [("PPA", ppa_known), ("WC", wc_known)]:
    print(f"\n{name}: net earned premium ($000), zero-paid company-lines")
    print(zero_paid_premium(known).sort_values("total_prem").to_string())
```

### 6.3 Follow-up: losses. check if companies with no payment activity incurred losses at latest valuation.

```python
def zero_paid_incurred(known):
    # Probe 6 mask: True = company-line has no nonzero paid cell
    mask = (known.CumPaidLoss != 0).groupby(known.GRCODE).sum() == 0
    flagged = mask[mask].index

    # One row per company-AY, at its latest known lag
    latest = (known[known.GRCODE.isin(flagged)]
              .sort_values("DevelopmentLag")
              .drop_duplicates(["GRCODE", "AccidentYear"], keep="last"))

    return latest.groupby("GRCODE")[["IncurredLosses", "CumPaidLoss", "EarnedPremNet"]].sum()

for name, known in [("PPA", ppa_known), ("WC", wc_known)]:
    print(f"\n{name}: incurred, paid and premium ($000), zero-paid company-lines")
    print(zero_paid_incurred(known).sort_values("EarnedPremNet").to_string())
```

### 6.4 Follow-up: checking for nonzero incurred loss cells among the zero paid, zero incurred loss at latest valuation companies.

```python
# Nonzero incurred cells among the zero-paid company-lines
for name, known, zp in [("PPA", ppa_known, ppa_zeropaid), ("WC", wc_known, wc_zeropaid)]:
    cells = known[known.GRCODE.isin(zp.index) & (known.IncurredLosses != 0)]
    print(f"\n{name}: nonzero incurred cells, zero-paid company-lines")
    print(cells[["GRCODE", "AccidentYear", "DevelopmentLag", "IncurredLosses", "BulkLoss"]].to_string(index=False))
```

Finding: All 8 PPA and 26 WC zero-paid company-lines have incurred = 0 at the latest known valuation. 5 WC lines show small incurred amounts (≤ 70, $000) in earlier cells that net to 0 by 2007. 17 WC lines earn positive premium but record no losses, so the line’s losses aren’t in this data. No method can run on these lines. These results validate row 1 of the data cleaning rules table.

### 5.2 Losses in flagged AYs: check if business was actually performed in flagged AYs

```python
#PPA
def check52(known, flagged):
    keys = ["GRCODE", "AccidentYear"]
    # Latest known lag per company-AY (losses are cumulative)
    latest = known.loc[known.groupby(keys).DevelopmentLag.idxmax(), keys + ["DevelopmentLag", "CumPaidLoss", "IncurredLosses", "BulkLoss"]]
    latest["Reported"] = latest.IncurredLosses - latest.BulkLoss
    out = flagged.merge(latest[keys + ["DevelopmentLag", "CumPaidLoss", "Reported"]], on=keys, how="left")
    out["no_losses"] = (out.CumPaidLoss == 0) & (out.Reported == 0)

    lines = out.groupby("GRCODE").no_losses.all()
    print(f"Flagged company-AYs: {len(out)} | no losses: {out.no_losses.sum()} | with losses: {(~out.no_losses).sum()}")
    print(f"Flagged company-lines: {len(lines)} | all flagged AYs loss-free: {lines.sum()} | ≥1 flagged AY with losses: {(~lines).sum()}")
    print(out[~out.no_losses].sort_values(keys).to_string(index=False))
    return out

ppa_52 = check52(ppa_known, ppa_p5)
```

```python
#WC
def check52(known, flagged):
    keys = ["GRCODE", "AccidentYear"]
    # Latest known lag per company-AY (losses are cumulative)
    latest = known.loc[known.groupby(keys).DevelopmentLag.idxmax(), keys + ["DevelopmentLag", "CumPaidLoss", "IncurredLosses", "BulkLoss"]]
    latest["Reported"] = latest.IncurredLosses - latest.BulkLoss
    out = flagged.merge(latest[keys + ["DevelopmentLag", "CumPaidLoss", "Reported"]], on=keys, how="left")
    out["no_losses"] = (out.CumPaidLoss == 0) & (out.Reported == 0)

    lines = out.groupby("GRCODE").no_losses.all()
    print(f"Flagged company-AYs: {len(out)} | no losses: {out.no_losses.sum()} | with losses: {(~out.no_losses).sum()}")
    print(f"Flagged company-lines: {len(lines)} | all flagged AYs loss-free: {lines.sum()} | ≥1 flagged AY with losses: {(~lines).sum()}")
    print(out[~out.no_losses].sort_values(keys).to_string(index=False))
    return out

wc_52 = check52(wc_known, wc_p5)
```

## Check 7. ceded-to-direct premium ratio swings
Question: What does the difference between the largest and smallest ratio per company line look like and can a rule be constructed and validated with this data?

### 7.1 ceded ÷ direct swing, with input checks
Prints, for company-lines left after rules 1–2:
- Line and company-AY counts
- Identity check: net = direct − ceded in every AY
- AYs with direct = 0, and AYs with a ratio < 0 or ≥ 1
- Lines with < 3 AYs (too few for a meaningful swing)
- Swing distribution (median, p75, p90, max) and lines with swing > 0.25 and > 0.5

```python
# PPA
def probe7(known, excluded):
    ay = (known[~known.GRCODE.isin(excluded)]
          .drop_duplicates(["GRCODE", "AccidentYear"])
          [["GRCODE", "AccidentYear", "EarnedPremDIR", "EarnedPremCeded", "EarnedPremNet"]]
          .copy())
    ay["ratio"] = (ay.EarnedPremCeded / ay.EarnedPremDIR).where(ay.EarnedPremDIR != 0)
    g = ay.groupby("GRCODE")
    line = pd.DataFrame({"n_ay": g.size(), "swing": g.ratio.max() - g.ratio.min()})

    print("lines:", len(line), "| company-AYs:", len(ay))
    print("net = direct - ceded:",
          ((ay.EarnedPremDIR - ay.EarnedPremCeded - ay.EarnedPremNet).abs() < 1).all())
    print("direct = 0 AYs:", (ay.EarnedPremDIR == 0).sum(),
          "| ratio < 0 or >= 1 AYs:", ((ay.ratio < 0) | (ay.ratio >= 1)).sum())
    print("lines with < 3 AYs (swing not meaningful):", (line.n_ay < 3).sum())
    print(line.swing.describe(percentiles=[.5, .75, .9])
          .loc[["count", "50%", "75%", "90%", "max"]].round(3))
    for t in (0.25, 0.5):
        print(f"swing > {t}:", (line.swing > t).sum())
    return ay, line

excluded = set(ppa_zeropaid.index) | set(ppa_p5.GRCODE.unique())
ppa_p7_ay, ppa_p7 = probe7(ppa_known, excluded)
```

```python
# WC
def probe7(known, excluded):
    ay = (known[~known.GRCODE.isin(excluded)]
          .drop_duplicates(["GRCODE", "AccidentYear"])
          [["GRCODE", "AccidentYear", "EarnedPremDIR", "EarnedPremCeded", "EarnedPremNet"]]
          .copy())
    ay["ratio"] = (ay.EarnedPremCeded / ay.EarnedPremDIR).where(ay.EarnedPremDIR != 0)
    g = ay.groupby("GRCODE")
    line = pd.DataFrame({"n_ay": g.size(), "swing": g.ratio.max() - g.ratio.min()})

    print("lines:", len(line), "| company-AYs:", len(ay))
    print("net = direct - ceded:",
          ((ay.EarnedPremDIR - ay.EarnedPremCeded - ay.EarnedPremNet).abs() < 1).all())
    print("direct = 0 AYs:", (ay.EarnedPremDIR == 0).sum(),
          "| ratio < 0 or >= 1 AYs:", ((ay.ratio < 0) | (ay.ratio >= 1)).sum())
    print("lines with < 3 AYs (swing not meaningful):", (line.n_ay < 3).sum())
    print(line.swing.describe(percentiles=[.5, .75, .9])
          .loc[["count", "50%", "75%", "90%", "max"]].round(3))
    for t in (0.25, 0.5):
        print(f"swing > {t}:", (line.swing > t).sum())
    return ay, line

excluded = set(wc_zeropaid.index) | set(wc_p5.GRCODE.unique())
wc_p7_ay, wc_p7 = probe7(wc_known, excluded)
```

### 7.2 Follow-up: AYs where net ≠ direct - ceded or the ratio is less than 0 or greater than or equal to 1. 7.1 turned up strange results so the premium data needs to be verified as sound before proceeding.

```python
# PPA
def p7_check(ay):
    ay = ay.assign(resid=ay.EarnedPremDIR - ay.EarnedPremCeded - ay.EarnedPremNet)
    print("resid value counts:", ay.resid.value_counts().sort_index().to_dict())
    bad_ratio = (ay.EarnedPremDIR == 0) | (ay.ratio < 0) | (ay.ratio >= 1)
    print("ratio out of range: AYs", bad_ratio.sum(), "| lines", ay[bad_ratio].GRCODE.nunique())
    out = ay[bad_ratio].sort_values(["GRCODE", "AccidentYear"])
    print(out.to_string())
    return out

ppa_p7_bad = p7_check(ppa_p7_ay)
```

```python
# WC
def p7_check(ay):
    ay = ay.assign(resid=ay.EarnedPremDIR - ay.EarnedPremCeded - ay.EarnedPremNet)
    print("resid value counts:", ay.resid.value_counts().sort_index().to_dict())
    bad_ratio = (ay.EarnedPremDIR == 0) | (ay.ratio < 0) | (ay.ratio >= 1)
    print("ratio out of range: AYs", bad_ratio.sum(), "| lines", ay[bad_ratio].GRCODE.nunique())
    out = ay[bad_ratio].sort_values(["GRCODE", "AccidentYear"])
    print(out.to_string())
    return out

wc_p7_bad = p7_check(wc_p7_ay)
```

### 9.2 Follow-up: is the answer key complete, and why are AYs missing?
This is the only check that reads hidden cells. It asks whether a grade exists, not what the grade would be, so it doesn't leak results into the cleaning rules.

Prints:

- For context: cells where cumulative paid decreases, hidden vs known. Small decreases are normal (salvage and subrogation recoveries).
- Zero rows in the full file: company-AYs with net premium = 0 and paid and incurred = 0 at every lag. If the file keeps AYs with no business as zero rows, a missing AY isn't an AY with no business.

```python
def p9_check(full, excluded):
    d = full[~full.GRCODE.isin(excluded)].sort_values(["GRCODE", "AccidentYear", "DevelopmentLag"])
    hidden = d.AccidentYear + d.DevelopmentLag - 1 > 2007
    prev = d.groupby(["GRCODE", "AccidentYear"]).CumPaidLoss.shift()
    blank = d[["CumPaidLoss", "IncurredLosses", "BulkLoss"]].isna().any(axis=1)
    to_zero = (d.CumPaidLoss == 0) & (prev > 0)
    print("hidden cells:", hidden.sum(), "| blank:", (hidden & blank).sum(),
          "| paid falls to 0:", (hidden & to_zero).sum())
    print(d[hidden & to_zero][["GRCODE", "AccidentYear", "DevelopmentLag", "CumPaidLoss", "EarnedPremNet"]]
          .to_string(index=False))
    dec = d.CumPaidLoss < prev
    print("paid decreases: hidden", (hidden & dec).sum(), "of", hidden.sum(),
          "| known", (~hidden & dec).sum(), "of", (~hidden).sum())
    g = full.groupby(["GRCODE", "AccidentYear"])
    zero_row = (g.EarnedPremNet.max() == 0) & (g.CumPaidLoss.apply(lambda s: (s == 0).all())) \
               & (g.IncurredLosses.apply(lambda s: (s == 0).all()))
    print("zero rows in full file: company-AYs", zero_row.sum(),
          "| lines", zero_row[zero_row].index.get_level_values("GRCODE").nunique())

for name, full, excl in [("PPA", ppa, ppa_excl), ("WC", wc, wc_excl)]:
    print(f"\n{name}")
    p9_check(full, excl)
```

### 8.2 Follow-up: development noise by size tercile
For each line, the paid 12→24 month factor for every AY with both lags known (AYs 1998–2006) and paid at 12 months > 0. Noise = coefficient of variation (std ÷ mean) of that factor across the line's AYs. Lines with fewer than 3 factors get no CV. Prints the median CV and median factor by size tercile. Known cells only.

```python
def p8_check(known, size):
    k = known[known.GRCODE.isin(size.index) & known.DevelopmentLag.isin([1, 2])]
    w = k.pivot_table(index=["GRCODE", "AccidentYear"], columns="DevelopmentLag",
                      values="CumPaidLoss", aggfunc="first").dropna()
    w = w[w[1] > 0]
    f = (w[2] / w[1]).groupby("GRCODE")
    line = pd.DataFrame({"n_f": f.size(), "factor": f.median(), "cv": f.std() / f.mean()})
    line.loc[line.n_f < 3, "cv"] = np.nan
    line["tercile"] = pd.qcut(size.reindex(line.index), 3, labels=["small", "mid", "large"])
    print("lines with no 12-24 factor:", len(size) - len(line))
    print(line.groupby("tercile", observed=True)
              .agg(lines=("cv", "size"), with_cv=("cv", "count"), median_cv=("cv", "median"),
                   median_factor=("factor", "median"))
              .to_string(float_format="{:.3f}".format))
    return line

for name, known, size in [("PPA", ppa_known, ppa_p8), ("WC", wc_known, wc_p8)]:
    print(f"\n{name}: paid 12-24 factor by size tercile")
    globals()[f"{name.lower()}_p8_cv"] = p8_check(known, size)
```

Result: Development noise rises as size falls. Median CV of the paid 12→24 factor, small / mid / large: PPA 0.124 / 0.090 / 0.053; WC 0.235 / 0.160 / 0.067. Small lines are 2–3.5× noisier than large ones. Median factors are similar across terciles (PPA 1.57 / 1.74 / 1.71, WC 2.12 / 2.29 / 2.23), so size changes the noise in the development pattern, not the pattern itself. Lines without a CV: PPA 2 small lines with fewer than 3 factors; WC 3 lines with no 12→24 factor and 1 with fewer than 3.

Finding: No drop. Size breaks no method: a small line's factors are noisier but still defined, and how much that noise hurts each method is something the backtest is supposed to measure. A cutoff at $1M per AY would remove 10 WC lines (one sixth), and sizes rise smoothly through $1M, so there is no break to put a cutoff in (unlike rule 4's cutoff). Size uses known premium only, so in Phase 4 it can be used to split results without leakage: report errors by size tercile. Small lines can have near-zero actual unpaid, which makes error ÷ actual unpaid explode, so errors are also reported as % of earned premium (closes the error-scale decision as "both").
